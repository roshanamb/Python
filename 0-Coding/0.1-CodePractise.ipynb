```python
## Write a program to print suqare of given number.
def generate_square(num):
    squ=[];
    for i in range(num):
        squ.append('*' * num)
    return squ;

lst = generate_square(5);
print('\n'.join(lst));
```

```python
## Write a program to print hollow square of given number.
def generate_hollow_square(num):
    squ=[];
    for i in range(num):
        if i == 0 or i == num - 1:
            squ.append('*' * num)
        else:
            squ.append('*' + ' ' * (num - 2) + '*')
    return squ;

lst = generate_hollow_square(5);
print('\n'.join(lst));
```

```python
## Write a program to print rectangle of given number.
def generate_rectangle(n,m):
    squ=[];
    for i in range(m):
        squ.append('*' * n)
    return squ;

lst = generate_rectangle(5, 3);
print('\n'.join(lst));
```

```python
## Write a program to print right angle triangle of given number.
def generate_right_angle_triangle(n):
    squ=[];
    for i in range(n):
        squ.append('*' * (i + 1))
    return squ;

def generate_inverted_triangle(n):
    squ=[];
    for i in range(n, 0, -1):
        squ.append('*' * i)
    return squ;

print('\n'.join(generate_right_angle_triangle(5)));
print('\n'.join(generate_inverted_triangle(8)));
```

```python
def generate_pyramid(n):
    squ=[];
    for i in range(n):
        squ.append(' ' * (n - i - 1) + '*' * (2 * i + 1) + ' ' * (n - i - 1))
    return squ;

print('\n'.join(generate_pyramid(5)));
```

```python
def generate_floyds_triangle(n):
    num = 1
    for i in range(n):
        for j in range(i + 1):
            print(num, end=' ')
            num += 1
        print()


lst = generate_floyds_triangle(5)
```

```python
##largest element in a list
numbers = [10, 24, 56, 78, 34, 89]
largest = numbers[0]
for num in numbers:
    if num > largest:
        largest = num

largest = max(numbers)

print("Largest element is:", largest)
print("Largest element is using MAX:", largest)
```

```python
##remove duplicates from a list
def remove_duplicates(lst):
    unique_lst = []
    for item in lst:
        if item not in unique_lst:
            unique_lst.append(item)
    return unique_lst
my_list = [1, 2, 2, 3, 4, 4, 5]
new_list = remove_duplicates(my_list)
print("List after removing duplicates:", new_list)
print("Using SET function:", list(set(my_list)))
```

```python
## count and returns the number of even and odd numbers in the list as tuple

numbers = [1,2,3,4,5]
even_count = 0
odd_count = 0
for num in numbers:
    if num % 2 == 0:
        even_count += 1
    else:
        odd_count += 1

print("Even and Odd count as tuple:", (even_count, odd_count))

##use map & lambda to find even & odd counts
numbers = [1,2,3,4,5]
even_count = len(list(filter(lambda x: x % 2 == 0, numbers)))
odd_count = len(list(filter(lambda x: x % 2 != 0, numbers)))
print("Even and Odd count as tuple using map & lambda:", (even_count, odd_count))
```

```python
##Merge two Sorted List
list1 = [1, 3, 5, 7]
list2 = [2, 4, 6, 8]
merged_list = []
i, j = 0, 0
while i < len(list1) and j < len(list2):
    if list1[i] < list2[j]:
        merged_list.append(list1[i])
        i += 1
    else:
        merged_list.append(list2[j])
        j += 1
merged_list.extend(list1[i:])
merged_list.extend(list2[j:])
print("Merged sorted list:", merged_list)
```
