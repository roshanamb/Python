```python
## Reverse a string
def reverse_string(s):
    return s[::-1]

print(reverse_string("Hello, World!"))  # Output: !dlroW ,olleH

## Count Vowels in a string
def count_vowels(s):
    vowels = 'aeiouAEIOU'
    return sum(1 for char in s if char in vowels)

print(count_vowels("Hello, World!"))  # Output: 3

## Count Consonants in a string
def count_consonants(s):
    vowels = 'aeiouAEIOU'
    return sum(1 for char in s if char.isalpha() and char not in vowels)

print(count_consonants("Hello, World!"))  # Output: 7

## Function to check if two strings are equal without using built-in functions.
def are_strings_equal(s1, s2):
    return s1 == s2
print(are_strings_equal("test", "test"))  # Output: True
print(are_strings_equal("test", "Test"))  # Output: False
```

```python
def are_is_palindrom(s):
    return s == s[::-1]
print(are_is_palindrom("racecar"))  # Output: True
print(are_is_palindrom("hello"))    # Output: False

## Word count in a string
def word_count(s):
    words = s.split()
    return len(words)
print(word_count("Hello, World! This is a test."))  # Output: 6

## Function to remove duplicate characters from the input string.
def remove_duplicates(s):
    seen = set()
    result = []
    for char in s:
        if char not in seen:
            seen.add(char)
            result.append(char)
    return ''.join(result)
print(remove_duplicates("banana"))  # Output: ban


```

```python
## Function to check if t is an anagram of s.
def are_anagrams(s, t):
    return sorted(s) == sorted(t)
print(are_anagrams("listen", "silent"))  # Output: True

## Function to check if t is an anagram of s without using built-in functions.
def are_anagrams_no_builtin(s, t):
    if len(s) != len(t):
        return False
    count = {}
    for char in s:
        count[char] = count.get(char, 0) + 1
    for char in t:
        if char not in count or count[char] == 0:
            return False
        count[char] -= 1
    return True
print(are_anagrams_no_builtin("listen", "silent"))  # Output: True
```

```python
## Function to check if t is a subsequence of s.
def is_subsequence(s, t):
    it = iter(s)
    return all(char in it for char in t)
print(is_subsequence("abcdef", "ace"))  # Output: True
print(is_subsequence("abcdef", "aec"))  # Output: False
```

```python
## Function to check if string t is a substring of string s
def is_substring(s, t):
    return t in s
print(is_substring("Hello, World!", "World"))  # Output: True
print(is_substring("Hello, World!", "world"))  # Output: False

## Function to check if string t is a substring of string s without using built-in functions.
def is_substring_no_builtin(s, t):
    len_s, len_t = len(s), len(t)
    for i in range(len_s - len_t + 1):
        if s[i:i+len_t] == t:
            return True
    return False
print(is_substring_no_builtin("Hello, World!", "World"))  # Output: True
print(is_substring_no_builtin("Hello, World!", "world"))  # Output: False
```

```python
## Function to find the length of the longest word in a string
def longest_word_length(s):
    words = s.split()
    return max(len(word) for word in words) if words else 0
print(longest_word_length("Hello, World! This is a test."))  # Output: 5
print(longest_word_length(""))  # Output: 0
```
