```python
# Celcius to Farenheit converter
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit

# Area of a rectangle
def area_of_rectangle(length, width):
    area = length * width
    return area

# Calculate distance traveled given speed and time
def calculate_distance(speed, time):
    distance = speed * time
    return distance

import math
math.ceil(4.2)  # Example usage of math.ceil function

print("25 Celsius is", celsius_to_fahrenheit(25), "Fahrenheit")  # Example usage
print("Area of rectangle with length 5 and width 3 is", area_of_rectangle(5, 3))     # Example usage
print("Distance traveled at 60 km/h for 2 hours is", calculate_distance(60, 2), "km")   # Example usage
```

```python
# Sum of n even natural numbers
def sum_of_even_natural_numbers(n):
    sum_even = n * (n + 1)
    return sum_even

print("Sum of first 5 even natural numbers is", sum_of_even_natural_numbers(5))  # Example usage


def is_even(number):
    return number % 2 == 0

print("Is 10 even?", is_even(10))  # Example usage
print("Is 7 even?", is_even(7))    # Example usage
```

```python
import math

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n/2) + 1):
        if n % i == 0:
            return False
    return True

def is_perfect_square(n):
    # return int(math.sqrt(n)) ** 2 == n
    # solve without using math.sqrt
    x = n // 2
    while x * x > n:
        x = (x + n // x) // 2
    return x * x == n

print("Is 29 prime?", is_prime(29))  # Example usage
print("Is 16 a perfect square?", is_perfect_square(16))  # Example usage
```

```python
def greatest_common_divisor(a, b):
    while b:
        a, b = b, a % b
    return a

print("GCD of 48 and 18 is", greatest_common_divisor(48, 18))  # Example usage

def least_common_multiple(a, b):
    return abs(a * b) // greatest_common_divisor(a, b)

print("LCM of 4 and 6 is", least_common_multiple(4, 6))  # Example usage
```
