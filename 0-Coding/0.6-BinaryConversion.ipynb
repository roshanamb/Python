```python
def int_to_binary(n):
    # return bin(n)[2:]
    if n == 0:
        return "0"

    binary = ""

    isNegative = n < 0
    if isNegative:
        n = -n # Handle negative numbers by converting to positive and adding a negative sign later

    while n > 0:
        binary = str(n % 2) + binary
        print(f"Current n: {n}, binary: {binary}")  # Debug statement to trace the computation
        n //= 2

    if isNegative:
        binary = "-" + binary  # Add negative sign back for negative numbers
        
    return binary

print("Binary representation of 10 is", int_to_binary(10))  # Example usage
```

```python
def binary_to_int(binary_str):
    #return int(binary_str, 2)
    result = 0
    for i, digit in enumerate(binary_str[::-1]):  # Process digits from right to left
        result = result + int(digit) * (2 ** i)  # Calculate the value of each digit based on its position
        print(f"Current digit: {digit}, result: {result} - {result} + {digit}*2**{i}")  # Debug statement to trace the computation
    return result

#10101 = 1*2^4 + 0*2^3 + 1*2^2 + 0*2^1 + 1*2^0 = 16 + 0 + 4 + 0 + 1 = 21
print("Binary 11101 is", binary_to_int("11101"))
```
